Funciones

```python
def dibujar_mexico(ax):
    mexico = geopandas.read_file("https://gist.githubusercontent.com/walkerke/76cb8cc5f949432f9555/raw/363c297ce82a4dcb9bdf003d82aa4f64bc695cf1/mx.geojson")
    mexico.plot(ax=ax)

def dibujar_capitales(capitales, ax):
    geometry = [Point(xy) for xy in zip(capitales['Longitud'], capitales['Latitud'])]
    geo_df = geopandas.GeoDataFrame(capitales, geometry = geometry)
    geo_df.plot(ax=ax,color='red', marker='o', label='ciudad')

def dibujar_recorrido(perm, ax):
    df_temp = capitales.reindex(perm)
    return ax.plot(df_temp.Longitud, df_temp.Latitud, '-k')

def tiempo_ciudades(ciudad1, ciudad2):
    return tiempo.at[ciudad1, ciudad2]

def tiempo_total(ruta):
    distancia = 0
    for i in range(32):
        ciudad_actual = ruta[i]
        ciudad_siguiente = ruta[(i + 1) % 32]
        distancia += tiempo_ciudades(ciudad_actual, ciudad_siguiente)
    return distancia
```

Leer Datos

```python
capitales = pd.read_csv('Capitales.csv')
capitales.set_index('Capital', inplace=True)
tiempo = pd.read_csv('tiempo_transporte2.csv')
tiempo.set_index('Origen', inplace=True)
```

Ejemplo

```python
# Crear alguna figura
fig, ax = plt.subplots(figsize=(10,5))

# Dibujar México
dibujar_mexico(ax)

# Dibujar capitales
dibujar_capitales(capitales, ax)

# Elegir una permutación
perm = random.sample(list(capitales.index), 32)
dibujar_recorrido(perm, ax)
ax.set_title(f'Tiempo: {tiempo_total(perm):.2f} horas', fontsize=12)
plt.show()
```
